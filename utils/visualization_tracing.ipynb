```python
import os
import pickle
import torch
import matplotlib.pyplot as plt
```

```python
file_path = '/data0/fangjunjie/federal_au/exp_results/multvae/ml-1m/adv/weights-attribute_gender-acall_grad=600/'
```

```python
with open(os.path.join(file_path, 'trace.pkl'), 'rb') as f:
    tracing_dict = pickle.load(f)
```

```python
for key, values in tracing_dict.items():
    values = [value.cpu().detach().numpy() if isinstance(value, torch.Tensor) else value for value in values]
    plt.figure()
    plt.plot(values, label=key)
    plt.legend()
    plt.xlabel('epochs')
    plt.ylabel(key)
    plt.savefig(os.path.join(file_path, f'tracing_{key}.png'))
    plt.show()
    plt.close()
```
